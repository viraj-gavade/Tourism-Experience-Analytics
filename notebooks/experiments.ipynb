## 3. ***Data Wrangling***

### Data Wrangling Code

#### Handling Missing Values 

````python
df['CityName'].fillna(df['CityName'].mode()[0], inplace=True)
````

#### Removing the duplicates 

````python
# removing the duplicate values if any
df = df.drop_duplicates()
````

#### Fixing the types 

````python
# Fixing the types 


from sklearn.preprocessing import LabelEncoder, StandardScaler



df['VisitYear'] = df['VisitYear'].astype(int)
df['VisitMonth'] = df['VisitMonth'].astype(int)
df['Rating'] = df['Rating'].astype(float)


text_cols = [
    'CityName', 'Country', 'Region', 'Continent',
    'Attraction', 'AttractionAddress', 'AttractionType',
    'VisitMode'
]

for col in text_cols:
    df[col] = df[col].str.strip()
    df[col] = df[col].str.lower()




print("Shape:", df.shape)
print("Missing Values:\n", df.isnull().sum())
````

## ✅ Data Manipulations Performed

To make the dataset clean and analysis-ready, the following data wrangling steps were performed:

## ***4. Data Vizualization, Storytelling & Experimenting with charts : Understand the relationships between variables***

#### Chart - 1

````python
# Chart - 1  Rating Distribution
plt.figure(figsize=(10,6))
sns.histplot(df['Rating'], bins=30, kde=True, color='blue')
plt.title('Rating Distribution')
plt.xlabel('Rating')
plt.ylabel('Frequency')
plt.show()
````

#### Chart - 2

````python
# Chart - 2 Visit Mode Distribution
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='VisitMode', palette='Set2')
plt.title('Visit Mode Distribution')
plt.xlabel('Visit Mode')
plt.ylabel('Count')
plt.show()
````

#### Chart - 3

````python
# Chart - 3 users by Continent
plt.figure(figsize=(10,6))
sns.countplot(data=df, x='Continent', order=df['Continent'].value_counts().index, palette='Set3')
plt.title('Users by Continent')
plt.xlabel('Continent')
plt.ylabel('Count')
plt.show()
````

#### Chart - 4

````python
# Chart - 4 Users by country (Top 10)
top_countries = df['Country'].value_counts().nlargest(10).index
plt.figure(figsize=(21,6))
sns.countplot(data=df[df['Country'].isin(top_countries)], x='Country', order=top_countries, palette='Set1')
plt.title('Top 10 Users by Country')
plt.xlabel('Country')
plt.ylabel('Count')
plt.show()
````

````python
# Chart - 5 Popular Attraction Types
plt.figure(figsize=(21,6))
top_attraction_types = df['AttractionType'].value_counts().nlargest(10).index
sns.countplot(data=df[df['AttractionType'].isin(top_attraction_types)], x='AttractionType', order=top_attraction_types)
plt.title('Top 10 Popular Attraction Types')
plt.xlabel('Attraction Type')
plt.ylabel('Count')
plt.show()
````

````python
# Chart - 6 Average Rating by atrraction Type
avg_rating_by_attraction = df.groupby('AttractionType')['Rating'].mean().sort_values(ascending=False).head(10)
plt.figure(figsize=(18,6))
sns.barplot(x=avg_rating_by_attraction.index, y=avg_rating_by_attraction.values, palette='Blues_d')
plt.title('Top 10 Attraction Types by Average Rating')
plt.xlabel('Attraction Type')
plt.ylabel('Average Rating')
plt.show()
````

#### Chart - 7

````python
# Chart - 7 Average Rating by Visit Mode
avg_rating_by_visit_mode = df.groupby('VisitMode')['Rating'].mean().sort_values(ascending=False)
plt.figure(figsize=(10,6))
sns.barplot(x=avg_rating_by_visit_mode.index, y=avg_rating_by_visit_mode.values, palette='Greens_d')
plt.title('Average Rating by Visit Mode')
plt.xlabel('Visit Mode')
plt.ylabel('Average Rating')
plt.show()
````

#### Chart - 8

````python
# Chart - 8 Ratings by Continent Box plot
plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Continent', y='Rating', palette='Pastel1')
plt.title('Ratings by Continent')
plt.xlabel('Continent')
plt.ylabel('Rating')
plt.show()
````

#### Chart - 9

````python
# Chart - 9 Rating by attraction Type Box plot
plt.figure(figsize=(21,6))
sns.boxplot(data=df[df['AttractionType'].isin(top_attraction_types)], x='AttractionType', y='Rating', palette='Set3')
plt.title('Ratings by Top 10 Attraction Types')
plt.xlabel('Attraction Type')
plt.ylabel('Rating')
plt.show()
````

#### Chart - 10

````python
# Chart - 10 Visit mode vs continent heatmap
visit_continent_pivot = df.pivot_table(index='VisitMode', columns='Continent', values='Rating', aggfunc='mean')
plt.figure(figsize=(10,6))
sns.heatmap(visit_continent_pivot, annot=True, fmt=".2f", cmap='YlGnBu')
plt.title('Average Rating by Visit Mode and Continent')
plt.xlabel('Continent')
plt.ylabel('Visit Mode')
plt.show()
````

#### Chart - 11

````python
# Chart - 11 Vsit Mode vs Attraction Type heatmap
visit_attraction_pivot = df.pivot_table(index='VisitMode', columns='AttractionType', values='Rating', aggfunc='mean')
plt.figure(figsize=(21,8))
sns.heatmap(visit_attraction_pivot, annot=True, fmt=".2f", cmap='coolwarm')
plt.title('Average Rating by Visit Mode and Attraction Type')
plt.xlabel('Attraction Type')
plt.ylabel('Visit Mode')
plt.show()
````

#### Chart - 12

````python
# Chart - 12 Monthly Visit Trends
monthly_visit_counts = df['VisitMonth'].value_counts().sort_index()
plt.figure(figsize=(10,6))
sns.lineplot(x=monthly_visit_counts.index, y=monthly_visit_counts.values, marker='o')
plt.title('Monthly Visit Trends')
plt.xlabel('Month')
plt.ylabel('Number of Visits')
plt.show()
````

````python
# Chart - 13 yearly Visit Trends
yearly_visit_counts = df['VisitYear'].value_counts().sort_index()
plt.figure(figsize=(10,6))
sns.lineplot(x=yearly_visit_counts.index, y=yearly_visit_counts.values, marker='o')
plt.title('Yearly Visit Trends')
plt.xlabel('Year')
plt.ylabel('Number of Visits')
plt.show()
````

#### Chart - 14 - Correlation Heatmap

````python
# Correlation Heatmap visualization code
numeric_cols = df.select_dtypes(include=[np.number]).columns
correlation_matrix = df[numeric_cols].corr()
plt.figure(figsize=(8,6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap')
plt.show()
````

#### Chart - 15 - Pair Plot

````python
# Pair Plot visualization code
sns.pairplot(df[numeric_cols], diag_kind='kde', corner=True)
plt.suptitle('Pair Plot of Numerical Features', y=1.02)
plt.show()
````
